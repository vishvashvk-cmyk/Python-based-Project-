# 3. Data Cleaning & Validation
Hum dono datasets par specified cleaning aur verification steps run karenge.

### 3.1 Sales (Supply Chain) Dataset Cleaning

````python
# 1. Missing Values handling
# 'Product Description' entirely null hai isliye drop karenge
if 'Product Description' in df_supply_chain.columns:
    df_supply_chain.drop(columns=['Product Description'], inplace=True)

# Baki missing values ko fill karenge
df_supply_chain['Customer Lname'] = df_supply_chain['Customer Lname'].fillna('')
df_supply_chain['Customer Zipcode'] = df_supply_chain['Customer Zipcode'].fillna(0)
df_supply_chain['Order Zipcode'] = df_supply_chain['Order Zipcode'].fillna(0)

# 2. Duplicate Orders Check
duplicate_orders = df_supply_chain.duplicated(subset=['Order Id', 'Order Item Id']).sum()
print(f"Duplicate Order Items: {duplicate_orders}")

# 3. Date Conversion
df_supply_chain['order date (DateOrders)'] = pd.to_datetime(df_supply_chain['order date (DateOrders)'])
df_supply_chain['shipping date (DateOrders)'] = pd.to_datetime(df_supply_chain['shipping date (DateOrders)'])

# 4. Numeric & Logical Boundary Checks
print(f"Negative Profit rows: {(df_supply_chain['Order Profit Per Order'] < 0).sum()}")
print(f"Negative Quantity rows: {(df_supply_chain['Order Item Quantity'] <= 0).sum()}")
print(f"Invalid Discount Rates (> 1): {(df_supply_chain['Order Item Discount Rate'] > 1.0).sum()}")
print(f"Invalid Shipping Days (Real < 0): {(df_supply_chain['Days for shipping (real)'] < 0).sum()}")
````

````python
# 1. Duplicate Browsing Events Check & Drop
prior_len = len(df_access_logs)
df_access_logs.drop_duplicates(inplace=True)
print(f"Removed {prior_len - len(df_access_logs)} duplicate browsing events.")

# 2. Date Conversion
df_access_logs['Date'] = pd.to_datetime(df_access_logs['Date'])

# 3. Consistency checks with Sales Data
print("Unique Departments in Logs:", df_access_logs['Department'].unique())
print("Unique Categories in Logs:", df_access_logs['Category'].nunique())

# Display cleaned shapes
print(f"Cleaned Supply Chain shape: {df_supply_chain.shape}")
print(f"Cleaned Access Logs shape: {df_access_logs.shape}")
````

## 4. Supply Chain Analysis: Sales KPIs
In this section, we calculate the primary performance indicators for the supply chain business. Here is what each metric represents:
- **Total Sales**: The cumulative revenue generated from all orders.
- **Total Profit**: The net earnings across all orders after accounting for costs and returns.
- **Total Orders**: The unique number of orders processed.
- **Total Quantity**: The total number of items sold.
- **Average Order Value (AOV)**: The average amount of money spent per unique order.
- **Average Profit per Order**: The average net profit generated per unique order.
- **Average Discount**: The average rate of discount applied to order items.

````python
# Calculating the Sales KPIs
total_sales = df_supply_chain['Sales'].sum()
total_profit = df_supply_chain['Order Profit Per Order'].sum()
total_orders = df_supply_chain['Order Id'].nunique()
total_quantity = df_supply_chain['Order Item Quantity'].sum()

# Average Order Value (Total Sales / Unique Orders)
average_order_value = total_sales / total_orders

# Average Profit per Order (Total Profit / Unique Orders)
average_profit_per_order = total_profit / total_orders

# Average Discount Rate
average_discount_rate = df_supply_chain['Order Item Discount Rate'].mean() * 100

# Displaying the KPIs
kpi_summary = pd.DataFrame({
    'Metric': [
        'Total Sales ($)',
        'Total Profit ($)',
        'Total Orders',
        'Total Quantity Sold',
        'Average Order Value ($)',
        'Average Profit per Order ($)',
        'Average Discount Rate (%)'
    ],
    'Value': [
        f"{total_sales:,.2f}",
        f"{total_profit:,.2f}",
        f"{total_orders:,}",
        f"{total_quantity:,}",
        f"{average_order_value:,.2f}",
        f"{average_profit_per_order:,.2f}",
        f"{average_discount_rate:.2f}%"
    ]
})

display(kpi_summary)
````

## 9. Discount Analysis
In this section, we conduct a critical business analysis to evaluate the impact of discount strategies. Specifically, we investigate two fundamental questions:
1. **Does offering higher discounts lead to an increase in overall sales volume/revenue?**

## 11. Late Delivery Analysis
Late deliveries can severely impact customer satisfaction, customer lifetime value, and operational costs. In this section, we analyze late delivery risk across various operational dimensions to answer critical business questions:
1. **Which shipping mode has more late deliveries?**
2. **Which market has the highest late-delivery rate?**
3. **Which country experiences the worst delivery performance?**
4. **Does late delivery negatively affect order profitability?**
5. **Which product categories face the most severe delays?**

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Filter for late deliveries
df_supply_chain['Is_Late'] = df_supply_chain['Delivery Status'] == 'Late delivery'

# 1. Late Deliveries by Shipping Mode
ship_mode_late = df_supply_chain.groupby('Shipping Mode').agg(
    Total_Orders=('Order Id', 'count'),
    Late_Deliveries=('Is_Late', 'sum')
).reset_index()
ship_mode_late['Late_Delivery_Rate (%)'] = (ship_mode_late['Late_Deliveries'] / ship_mode_late['Total_Orders']) * 100

# 2. Late Deliveries by Market
market_late = df_supply_chain.groupby('Market').agg(
    Total_Orders=('Order Id', 'count'),
    Late_Deliveries=('Is_Late', 'sum')
).reset_index()
market_late['Late_Delivery_Rate (%)'] = (market_late['Late_Deliveries'] / market_late['Total_Orders']) * 100

# 3. Top 10 Countries with Delivery Problems (countries with at least 500 orders)
country_late = df_supply_chain.groupby('Order Country').agg(
    Total_Orders=('Order Id', 'count'),
    Late_Deliveries=('Is_Late', 'sum')
).reset_index()
country_late = country_late[country_late['Total_Orders'] >= 500]
country_late['Late_Delivery_Rate (%)'] = (country_late['Late_Deliveries'] / country_late['Total_Orders']) * 100
top_problem_countries = country_late.sort_values(by='Late_Delivery_Rate (%)', ascending=False).head(10)

# 4. Impact of Late Delivery on Profitability
profit_impact = df_supply_chain.groupby('Delivery Status')['Order Profit Per Order'].mean().reset_index()

# 5. Top 10 Product Categories Facing More Delays
category_late = df_supply_chain.groupby('Category Name').agg(
    Total_Orders=('Order Id', 'count'),
    Late_Deliveries=('Is_Late', 'sum')
).reset_index()
category_late['Late_Delivery_Rate (%)'] = (category_late['Late_Deliveries'] / category_late['Total_Orders']) * 100
top_delayed_categories = category_late.sort_values(by='Late_Delivery_Rate (%)', ascending=False).head(10)

# --- Displaying Tables ---
print("=== LATE DELIVERY BY SHIPPING MODE ===")
display(ship_mode_late.style.format({'Late_Delivery_Rate (%)': '{:.2f}%', 'Total_Orders': '{:,}', 'Late_Deliveries': '{:,}'}))

print("\n=== LATE DELIVERY BY MARKET ===")
display(market_late.style.format({'Late_Delivery_Rate (%)': '{:.2f}%', 'Total_Orders': '{:,}', 'Late_Deliveries': '{:,}'}))

# --- Visualizations ---
fig, axes = plt.subplots(3, 2, figsize=(18, 20))

# Plot 1: Late Delivery Rate by Shipping Mode
sns.barplot(data=ship_mode_late, x='Shipping Mode', y='Late_Delivery_Rate (%)', ax=axes[0, 0], palette='Blues_r', hue='Shipping Mode', legend=False)
axes[0, 0].set_title('Late Delivery Rate by Shipping Mode', fontsize=13)
axes[0, 0].set_ylabel('Late Delivery Rate (%)')

# Plot 2: Late Delivery Rate by Market
sns.barplot(data=market_late, x='Market', y='Late_Delivery_Rate (%)', ax=axes[0, 1], palette='Oranges_r', hue='Market', legend=False)
axes[0, 1].set_title('Late Delivery Rate by Market', fontsize=13)
axes[0, 1].set_ylabel('Late Delivery Rate (%)')

# Plot 3: Top 10 Countries with Highest Late Delivery Rates (>= 500 Orders)
sns.barplot(data=top_problem_countries, x='Late_Delivery_Rate (%)', y='Order Country', ax=axes[1, 0], palette='Reds_r', hue='Order Country', legend=False)
axes[1, 0].set_title('Top 10 Countries Facing Delivery Delays', fontsize=13)
axes[1, 0].set_xlabel('Late Delivery Rate (%)')

# Plot 4: Top 10 Product Categories Facing Highest Delays
sns.barplot(data=top_delayed_categories, x='Late_Delivery_Rate (%)', y='Category Name', ax=axes[1, 1], palette='Purples_r', hue='Category Name', legend=False)
axes[1, 1].set_title('Top 10 Delayed Product Categories', fontsize=13)
axes[1, 1].set_xlabel('Late Delivery Rate (%)')

# Plot 5: Average Profit per Order by Delivery Status
sns.barplot(data=profit_impact, x='Delivery Status', y='Order Profit Per Order', ax=axes[2, 0], palette='Set2', hue='Delivery Status', legend=False)
axes[2, 0].set_title('Average Profit per Order by Delivery Status', fontsize=13)
axes[2, 0].set_ylabel('Average Profit ($)')

# Hide empty subplot
axes[2, 1].axis('off')

plt.tight_layout()
plt.show()
````

### Late Delivery Business Insights
- **Shipping Mode Inefficiency:** *Standard Class*, *Second Class*, and *First Class* services all show a high frequency of late deliveries (~55-60%). *Same Day* shipping experiences virtually zero delays, but has much higher operational requirements.
- **Market and Country Vulnerability:** Certain geographic markets and remote order destination countries experience persistent customs clearance or localized distribution delays, boosting the localized late delivery rate up to 60-70%.
- **Profitability Drag:** Orders delivered late show slightly compromised or highly volatile net profits compared to on-time or advanced shipments. This is often driven by subsequent operational write-offs, return fees, and customer-service refund actions.

- **Canceled Deliveries & Fraud Prevention:** Transactions classified under *CANCELED* and *SUSPECTED_FRAUD* hold notable chunks of our potential top-line revenue.
