```python
import pandas as pd

# 1. Excel file load karna
df = pd.read_excel('sales_dataset.xlsx')

# 2. Pehli 5 rows dekhna
print("--- FIRST 5 ROWS ---")
display(df.head())

# 3. Data ka Shape (Total Rows, Total Columns) dekhna
print("\n--- SHAPE (Rows, Columns) ---")
print(df.shape)

# 4. Saare Columns ke naam dekhna
print("\n--- COLUMNS ---")
print(df.columns)

# 5. Data Types aur Non-Null Counts (info) check karna
print("\n--- DATA INFO ---")
df.info()

# 6. Data Types ki list dekhna
print("\n--- DATA TYPES ---")
print(df.dtypes)

# 7. Numerical columns ki summary statistics dekhna
print("\n--- NUMERICAL DESCRIBE ---")
display(df.describe())

# 8. Saare columns (Numerical + Categorical) ki summary dekhna
print("\n--- OVERALL DESCRIBE (INCLUDE ALL) ---")
display(df.describe(include="all"))
```

```python
# 1. Missing Values Check karna
print("--- MISSING VALUES PER COLUMN ---")
print(df.isnull().sum())

# 2. Duplicate Orders (Poori row same ho) check karna
print("\n--- TOTAL DUPLICATE ROWS ---")
print(df.duplicated().sum())

# 3. Duplicate Order IDs check karna (Kya ek Order ID ek se zyada baar aa raha hai?)
print("\n--- DUPLICATE ORDER IDs ---")
print(df['Order ID'].duplicated().sum())

# 4. 'Unnamed: 22' column ki unique values aur unka count check karna
if 'Unnamed: 22' in df.columns:
    print("\n--- UNNAMED: 22 COLUMN CHECK ---")
    print(df['Unnamed: 22'].value_counts(dropna=False))

# 5. Specific Columns ke DataTypes aur unki unique values / anomalies check karna
print("\n--- DATA TYPES CHECK FOR TARGET COLUMNS ---")
target_cols = [
    'Order ID', 'Date', 'Status', 'Fulfilment', 'Sales Channel',
    'Category', 'Size', 'Qty', 'Amount', 'ship-city',
    'ship-state', 'ship-country', 'promotion-ids', 'B2B',
    'fulfilled-by', 'Courier Status'
]

for col in target_cols:
    if col in df.columns:
        print(f"Column: {col} | Type: {df[col].dtype} | Missing: {df[col].isnull().sum()}")

# 6. Category spelling/unique values check karna
if 'Category' in df.columns:
    print("\n--- CATEGORY UNIQUE VALUES (Spelling check) ---")
    print(df['Category'].unique())

# 7. State aur City mein inconsistencies (jaise spelling mistakes ya extra spaces) check karna
if 'ship-state' in df.columns:
    print("\n--- SHIP-STATE UNIQUE VALUES (Sample) ---")
    # State names ko lowercase karke trailing/leading spaces hata kar unique check karna
    states = df['ship-state'].dropna().astype(str).str.strip().str.upper().unique()
    print(f"Total Unique States: {len(states)}")
    print(sorted(states)[:15]) # Pehle 15 states dikhayein

if 'ship-city' in df.columns:
    print(f"\nTotal Unique Cities: {df['ship-city'].nunique()}")
```

```python
# 1. Total Orders (Total Rows in Dataset)
total_orders = len(df)

# 2. Unique Order IDs (Real unique transactions)
unique_orders = df['Order ID'].nunique()

# 3. Total Units Sold (Sum of Qty)
total_units_sold = df['Qty'].sum()

# 4. Total Revenue (Sum of Amount)
total_revenue = df['Amount'].sum()

# 5. Average Order Value (AOV) = Total Revenue / Total Orders
aov = total_revenue / total_orders

# 6. Average Units per Order = Total Units Sold / Total Orders
avg_units_per_order = total_units_sold / total_orders

# 7. Cancelled Orders count
# Status column mein 'Cancelled' kitni baar hai check karna
cancelled_orders = df[df['Status'].str.lower() == 'cancelled'].shape[0]

# 8. Cancellation Rate (%) = (Cancelled Orders / Total Orders) * 100
cancellation_rate = (cancelled_orders / total_orders) * 100

# Results Print Karna
print("========================================")
print("          SALES KPI METRICS             ")
print("========================================")
print(f"Total Rows / Total Orders   : {total_orders:,}")
print(f"Unique Order IDs            : {unique_orders:,}")
print(f"Total Units Sold (Qty Sum)  : {total_units_sold:,}")
print(f"Total Revenue (Amount Sum)  : ₹{total_revenue:,.2f}")
print(f"Average Order Value (AOV)   : ₹{aov:,.2f}")
print(f"Average Units per Order     : {avg_units_per_order:.2f}")
print(f"Cancelled Orders            : {cancelled_orders:,}")
print(f"Cancellation Rate           : {cancellation_rate:.2f}%")
print("========================================")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Date column ko Datetime mein convert karke Month, Month Number, Year nikalna
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year
df['Month_Num'] = df['Date'].dt.month
df['Month'] = df['Date'].dt.strftime('%B') # Month Name (e.g. April, May)

# 2. Monthly Aggregation (Orders, Revenue, Units Sold)
monthly_sales = df.groupby(['Month_Num', 'Month', 'Year']).agg(
    Total_Orders=('Order ID', 'count'),
    Total_Revenue=('Amount', 'sum'),
    Total_Units_Sold=('Qty', 'sum')
).reset_index().sort_values(by='Month_Num')

print("--- MONTHLY SALES SUMMARY ---")
display(monthly_sales)

# 3. Key Questions / Insights Calculation
highest_revenue_row = monthly_sales.loc[monthly_sales['Total_Revenue'].idxmax()]
lowest_revenue_row = monthly_sales.loc[monthly_sales['Total_Revenue'].idxmin()]

print("\n========================================")
print("         MONTHLY SALES INSIGHTS         ")
print("========================================")
print(f"Highest Sales Month : {highest_revenue_row['Month']} (Revenue: ₹{highest_revenue_row['Total_Revenue']:,.2f} | Orders: {highest_revenue_row['Total_Orders']:,})")
print(f"Lowest Sales Month  : {lowest_revenue_row['Month']} (Revenue: ₹{lowest_revenue_row['Total_Revenue']:,.2f} | Orders: {lowest_revenue_row['Total_Orders']:,})")

# Trend Analysis
if len(monthly_sales) > 1:
    first_month_rev = monthly_sales['Total_Revenue'].iloc[0]
    last_month_rev = monthly_sales['Total_Revenue'].iloc[-1]
    trend = "Increasing" if last_month_rev > first_month_rev else "Decreasing"
    print(f"Sales Trend          : {trend} over the period")

print("========================================\n")

# 4. Visualizations: Line Charts
plt.figure(figsize=(15, 5))

# Plot 1: Monthly Revenue Trend
plt.subplot(1, 3, 1)
sns.lineplot(data=monthly_sales, x='Month', y='Total_Revenue', marker='o', color='b', linewidth=2.5)
plt.title('Monthly Revenue Trend (₹)')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.6)

# Plot 2: Monthly Orders Trend
plt.subplot(1, 3, 2)
sns.lineplot(data=monthly_sales, x='Month', y='Total_Orders', marker='o', color='g', linewidth=2.5)
plt.title('Monthly Total Orders')
plt.xlabel('Month')
plt.ylabel('Total Orders')
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.6)

# Plot 3: Monthly Units Sold
plt.subplot(1, 3, 3)
sns.lineplot(data=monthly_sales, x='Month', y='Total_Units_Sold', marker='o', color='r', linewidth=2.5)
plt.title('Monthly Units Sold')
plt.xlabel('Month')
plt.ylabel('Units Sold')
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Size-wise Aggregation (Total Orders, Total Units, Revenue)
size_summary = df.groupby('Size').agg(
    Total_Orders=('Order ID', 'count'),
    Total_Units=('Qty', 'sum'),
    Total_Revenue=('Amount', 'sum')
).reset_index().sort_values(by='Total_Orders', ascending=False)

print("--- SIZE ANALYSIS SUMMARY ---")
display(size_summary)

# 2. Key Questions Insights
most_sold_size = size_summary.iloc[0]['Size']
print(f"\nMost Sold Size: {most_sold_size}")

# 3. Category × Size Cross-Analysis (Pivot Table for Orders Count)
category_size_pivot = df.pivot_table(
    index='Category',
    columns='Size',
    values='Order ID',
    aggfunc='count',
    fill_value=0
)

print("\n--- CATEGORY × SIZE CROSS-ANALYSIS (ORDERS) ---")
display(category_size_pivot)

# 4. Visualizations
plt.figure(figsize=(14, 6))

# Plot 1: Orders by Size Bar Chart
plt.subplot(1, 2, 1)
sns.barplot(data=size_summary, x='Size', y='Total_Orders', palette='crest')
plt.title('Total Orders by Size')
plt.xlabel('Size')
plt.ylabel('Total Orders')
plt.xticks(rotation=45)

# Plot 2: Category × Size Heatmap
plt.subplot(1, 2, 2)
sns.heatmap(category_size_pivot, annot=False, fmt='d', cmap='YlGnBu', cbar=True)
plt.title('Category × Size Demand (Heatmap)')
plt.xlabel('Size')
plt.ylabel('Category')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Channel-wise Basic Aggregation
fulfilment_summary = df.groupby('Fulfilment').agg(
    Total_Orders=('Order ID', 'count'),
    Total_Units_Sold=('Qty', 'sum'),
    Total_Revenue=('Amount', 'sum')
).reset_index()

# 2. Cancellation Rate by Fulfilment Method Calculate Karna
# Har Fulfilment method ke kitne orders 'Cancelled' hain check karein
cancelled_by_fulfilment = df[df['Status'].str.lower() == 'cancelled'].groupby('Fulfilment')['Order ID'].count().reset_index()
cancelled_by_fulfilment.rename(columns={'Order ID': 'Cancelled_Orders'}, inplace=True)

# Merge summary with cancelled orders
fulfilment_summary = pd.merge(fulfilment_summary, cancelled_by_fulfilment, on='Fulfilment', how='left').fillna(0)

# Cancellation Rate (%) formula: (Cancelled Orders / Total Orders) * 100
fulfilment_summary['Cancellation_Rate_%'] = (fulfilment_summary['Cancelled_Orders'] / fulfilment_summary['Total_Orders']) * 100

print("========================================")
print("         FULFILMENT SUMMARY             ")
print("========================================")
display(fulfilment_summary)

# 3. Key Questions Insights
max_orders_method = fulfilment_summary.loc[fulfilment_summary['Total_Orders'].idxmax()]['Fulfilment']
max_revenue_method = fulfilment_summary.loc[fulfilment_summary['Total_Revenue'].idxmax()]['Fulfilment']

print("\n========================================")
print("         FULFILMENT INSIGHTS            ")
print("========================================")
print(f"Which method handles more orders?   : {max_orders_method}")
print(f"Which method has higher revenue?     : {max_revenue_method}")

print("\nCancellation Rates:")
for idx, row in fulfilment_summary.iterrows():
    print(f"- {row['Fulfilment']}: {row['Cancellation_Rate_%']:.2f}% ({int(row['Cancelled_Orders'])} cancelled out of {int(row['Total_Orders'])})")
print("========================================\n")

# 4. Visualizations: Bar Charts
plt.figure(figsize=(15, 5))

# Plot 1: Total Orders Comparison
plt.subplot(1, 3, 1)
sns.barplot(data=fulfilment_summary, x='Fulfilment', y='Total_Orders', palette='viridis')
plt.title('Total Orders by Fulfilment')
plt.xlabel('Fulfilment Method')
plt.ylabel('Total Orders')

# Plot 2: Total Revenue Comparison
plt.subplot(1, 3, 2)
sns.barplot(data=fulfilment_summary, x='Fulfilment', y='Total_Revenue', palette='magma')
plt.title('Total Revenue (₹) by Fulfilment')
plt.xlabel('Fulfilment Method')
plt.ylabel('Total Revenue')

# Plot 3: Cancellation Rate Comparison
plt.subplot(1, 3, 3)
sns.barplot(data=fulfilment_summary, x='Fulfilment', y='Cancellation_Rate_%', palette='rocket')
plt.title('Cancellation Rate (%)')
plt.xlabel('Fulfilment Method')
plt.ylabel('Cancellation Rate (%)')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# ==========================================
# 1. BASIC STATISTICAL METRICS (Amount)
# ==========================================
# Null values ko ignore karke calculations karenge
mean_val = df['Amount'].mean()
median_val = df['Amount'].median()
min_val = df['Amount'].min()
max_val = df['Amount'].max()
std_val = df['Amount'].std()

print("========================================")
print("       ORDER VALUE (AMOUNT) STATS       ")
print("========================================")
print(f"Mean (Average)       : ₹{mean_val:.2f}")
print(f"Median (Middle val)  : ₹{median_val:.2f}")
print(f"Minimum Value        : ₹{min_val:.2f}")
print(f"Maximum Value        : ₹{max_val:.2f}")
print(f"Standard Deviation   : ₹{std_val:.2f}")
print("========================================\n")


# ==========================================
# 2. DYNAMIC SEGMENTATION BASED ON DISTRIBUTION
# ==========================================
# Hum 33rd aur 67th percentile nikalenge as thresholds (Data ko 3 barabar hisson me baatne ke liye)
low_threshold = df['Amount'].quantile(0.33)
high_threshold = df['Amount'].quantile(0.67)

print("--- DEFINED THRESHOLDS (Based on Quantiles) ---")
print(f"Low Value Segment    : Up to ₹{low_threshold:.2f}")
print(f"Medium Value Segment : ₹{low_threshold:.2f} to ₹{high_threshold:.2f}")
print(f"High Value Segment   : Above ₹{high_threshold:.2f}\n")

# Segment assign karne ke liye function
def categorize_order_value(amount):
    if pd.isna(amount):
        return 'Unknown'
    elif amount <= low_threshold:
        return 'Low Value'
    elif amount <= high_threshold:
        return 'Medium Value'
    else:
        return 'High Value'

# Naya column create karna
df['Order_Value_Segment'] = df['Amount'].apply(categorize_order_value)

# Segments ki summary nikalna (Unknown ko hatakar)
segment_summary = df[df['Order_Value_Segment'] != 'Unknown'].groupby('Order_Value_Segment').agg(
    Total_Orders=('Order ID', 'count'),
    Total_Revenue=('Amount', 'sum')
).reindex(['Low Value', 'Medium Value', 'High Value']).reset_index()

print("========================================")
print("        VALUE SEGMENTATION SUMMARY      ")
print("========================================")
display(segment_summary)


# ==========================================
# 3. VISUALIZATIONS
# ==========================================
plt.figure(figsize=(18, 5))

# Plot 1: Amount Distribution (Histogram)
plt.subplot(1, 3, 1)
sns.histplot(df['Amount'].dropna(), bins=50, kde=True, color='purple')
plt.title('Distribution of Order Amount')
plt.xlabel('Amount (₹)')
plt.ylabel('Frequency')
plt.xlim(0, df['Amount'].quantile(0.99)) # Outliers ko hide karne ke liye 99th percentile tak dikhayenge

# Plot 2: Segment vs Total Orders
plt.subplot(1, 3, 2)
sns.barplot(data=segment_summary, x='Order_Value_Segment', y='Total_Orders', palette='viridis')
plt.title('Total Orders by Value Segment')
plt.xlabel('Value Segment')
plt.ylabel('Total Orders')

# Plot 3: Segment vs Total Revenue
plt.subplot(1, 3, 3)
sns.barplot(data=segment_summary, x='Order_Value_Segment', y='Total_Revenue', palette='magma')
plt.title('Total Revenue by Value Segment (₹)')
plt.xlabel('Value Segment')
plt.ylabel('Total Revenue (₹)')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================
# 1. DATA PREPARATION FOR CANCELLATION
# ==========================================
# Create a binary column for Cancellation (1 if Cancelled, 0 otherwise)
df['Is_Cancelled'] = df['Status'].apply(lambda x: 1 if str(x).strip().lower() == 'cancelled' else 0)

# Ensure Date and Month are ready
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.strftime('%B')

# Ensure State is clean
df['ship-state-clean'] = df['ship-state'].dropna().astype(str).str.strip().str.upper()

# Ensure Promotion Used exists
df['Promotion_Used'] = df['promotion-ids'].apply(lambda x: 'No' if pd.isna(x) else 'Yes')

# ==========================================
# 2. OVERALL CANCELLATION SUMMARY
# ==========================================
total_orders = len(df)
total_cancelled = df['Is_Cancelled'].sum()
overall_cancellation_rate = (total_cancelled / total_orders) * 100

print("========================================")
print("      OVERALL CANCELLATION SUMMARY      ")
print("========================================")
print(f"Total Orders      : {total_orders:,}")
print(f"Total Cancelled   : {total_cancelled:,}")
print(f"Overall Rate      : {overall_cancellation_rate:.2f}%\n")


# ==========================================
# 3. DIMENSION-WISE CANCELLATION RATES
# ==========================================
# Helper function to group and calculate cancellation rate
def get_cancellation_stats(dimension):
    stats = df.groupby(dimension).agg(
        Total_Orders=('Order ID', 'count'),
        Cancelled_Orders=('Is_Cancelled', 'sum')
    ).reset_index()
    stats['Cancellation_Rate_%'] = (stats['Cancelled_Orders'] / stats['Total_Orders']) * 100

    # Filtering out groups with less than 50 orders to avoid skewed/random 100% cancellation rates
    stats = stats[stats['Total_Orders'] > 50]
    return stats.sort_values('Cancellation_Rate_%', ascending=False)

# Calculate across different dimensions
cat_cancel = get_cancellation_stats('Category')
state_cancel = get_cancellation_stats('ship-state-clean').head(10) # Top 10 states with highest cancellation rate
fulfilment_cancel = get_cancellation_stats('Fulfilment')
shipping_cancel = get_cancellation_stats('ship-service-level')
month_cancel = get_cancellation_stats('Month')
b2b_cancel = get_cancellation_stats('B2B')
promo_cancel = get_cancellation_stats('Promotion_Used')

print("--- CANCELLATION BY CATEGORY ---")
display(cat_cancel)

print("\n--- CANCELLATION BY FULFILMENT ---")
display(fulfilment_cancel)

print("\n--- CANCELLATION BY PROMOTION ---")
display(promo_cancel)


# ==========================================
# 4. VISUALIZATIONS (DASHBOARD)
# ==========================================
plt.figure(figsize=(18, 10))

# Plot 1: Category
plt.subplot(2, 3, 1)
sns.barplot(data=cat_cancel, x='Category', y='Cancellation_Rate_%', hue='Category', palette='Reds_r', legend=False)
plt.title('Cancellation Rate by Category')
plt.xticks(rotation=45)
plt.ylabel('Cancellation Rate (%)')

# Plot 2: Fulfilment
plt.subplot(2, 3, 2)
sns.barplot(data=fulfilment_cancel, x='Fulfilment', y='Cancellation_Rate_%', hue='Fulfilment', palette='Oranges', legend=False)
plt.title('Cancellation Rate by Fulfilment')

# Plot 3: Shipping Level
plt.subplot(2, 3, 3)
sns.barplot(data=shipping_cancel, x='ship-service-level', y='Cancellation_Rate_%', hue='ship-service-level', palette='Blues', legend=False)
plt.title('Cancellation Rate by Shipping Level')

# Plot 4: Month
plt.subplot(2, 3, 4)
sns.barplot(data=month_cancel, x='Month', y='Cancellation_Rate_%', hue='Month', palette='Purples', legend=False)
plt.title('Cancellation Rate by Month')

# Plot 5: B2B
plt.subplot(2, 3, 5)
sns.barplot(data=b2b_cancel, x='B2B', y='Cancellation_Rate_%', hue='B2B', palette='Greens', legend=False)
plt.title('Cancellation Rate by B2B')

# Plot 6: Promotion
plt.subplot(2, 3, 6)
sns.barplot(data=promo_cancel, x='Promotion_Used', y='Cancellation_Rate_%', hue='Promotion_Used', palette='coolwarm', legend=False)
plt.title('Cancellation Rate by Promotion')

plt.tight_layout()
plt.show()
```
