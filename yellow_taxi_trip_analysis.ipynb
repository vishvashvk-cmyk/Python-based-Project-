```python
df = df[df['trip_distance'] > 0]
```

```python
df = df[df['passenger_count'] > 0]
```

```python
df = df[(df['fare_amount'] > 0) & (df['total_amount'] > 0)]
```

```python
df = df.dropna(subset=['passenger_count', 'trip_distance', 'fare_amount', 'total_amount'])
```

```python
# Master Data Cleaning Code
df_cleaned = df[
    (df['passenger_count'] > 0) &
    (df['trip_distance'] > 0) &
    (df['fare_amount'] > 0) &
    (df['total_amount'] > 0) &
    (df['avg_speed_mph'] <= 80)
].copy()

# Recalculate derived columns to remove any remaining NaNs
df_cleaned['fare_per_mile'] = df_cleaned['fare_amount'] / df_cleaned['trip_distance']

print("Old Shape:", df.shape)
print("New Cleaned Shape:", df_cleaned.shape)
```

```python
# Basic Taxi KPIs Calculation
total_trips = len(df_cleaned)
total_revenue = df_cleaned['total_amount'].sum()
avg_fare = df_cleaned['fare_amount'].mean()
avg_distance = df_cleaned['trip_distance'].mean()
avg_duration = df_cleaned['trip_duration_min'].mean()
avg_tip = df_cleaned['tip_amount'].mean()
avg_speed = df_cleaned['avg_speed_mph'].mean()
total_tips = df_cleaned['tip_amount'].sum()
avg_revenue_per_trip = df_cleaned['total_amount'].mean()

# Output Display
print(f"Total Trips: {total_trips:,}")
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Average Fare: ${avg_fare:.2f}")
print(f"Average Trip Distance: {avg_distance:.2f} miles")
print(f"Average Trip Duration: {avg_duration:.2f} mins")
print(f"Average Tip: ${avg_tip:.2f}")
print(f"Average Speed: {avg_speed:.2f} mph")
print(f"Total Tips: ${total_tips:,.2f}")
print(f"Average Revenue per Trip: ${avg_revenue_per_trip:.2f}")
```

```python
# Payment Analysis Code
payment_summary = df_cleaned.groupby('payment_type_label').agg(
    total_trips=('total_amount', 'count'),
    total_revenue=('total_amount', 'sum'),
    avg_fare=('fare_amount', 'mean'),
    avg_tip=('tip_amount', 'mean')
).reset_index()

# Percentage calculations
payment_summary['trip_percentage'] = (payment_summary['total_trips'] / payment_summary['total_trips'].sum()) * 100
payment_summary['revenue_percentage'] = (payment_summary['total_revenue'] / payment_summary['total_revenue'].sum()) * 100

print(payment_summary.to_string(index=False))

# Credit Card vs Cash Tip Behavior Analysis
cc_tips = df_cleaned[df_cleaned['payment_type_label'] == 'Credit Card']['tip_amount']
cash_tips = df_cleaned[df_cleaned['payment_type_label'] == 'Cash']['tip_amount']

print("\n--- Tip Behavior ---")
print(f"Credit Card Average Tip: ${cc_tips.mean():.2f}")
print(f"Cash Average Tip: ${cash_tips.mean():.2f}")
```

# 6. Peak vs Off-Peak Analysis

```python
import pandas as pd

print("=" * 60)
print("6. PEAK VS OFF-PEAK ANALYSIS - METRICS & COMPARISON")
print("=" * 60)

# Summary Aggregation Table comparing Peak vs Off-Peak
peak_summary = (
    df_cleaned.groupby("peak_period")
    .agg(
        total_trips=("total_amount", "count"),
        trip_share_pct=(
            "total_amount",
            lambda x: (len(x) / len(df_cleaned)) * 100,
        ),
        total_revenue=("total_amount", "sum"),
        revenue_share_pct=(
            "total_amount",
            lambda x: (x.sum() / df_cleaned["total_amount"].sum()) * 100,
        ),
        avg_fare_amount=("fare_amount", "mean"),
        avg_total_amount=("total_amount", "mean"),
        avg_distance_miles=("trip_distance", "mean"),
        avg_duration_min=("trip_duration_min", "mean"),
        avg_speed_mph=("avg_speed_mph", "mean"),
        avg_tip_rate=("tip_rate", "mean"),
    )
    .reindex(["Peak", "Off-Peak"])
)

print("\nPeak vs Off-Peak Summary Table:")
print(peak_summary.round(2))
print()

# Specific Behavioral Insights
peak_trips = peak_summary.loc["Peak", "total_trips"]
peak_trips_pct = peak_summary.loc["Peak", "trip_share_pct"]
offpeak_trips = peak_summary.loc["Off-Peak", "total_trips"]

peak_speed = peak_summary.loc["Peak", "avg_speed_mph"]
offpeak_speed = peak_summary.loc["Off-Peak", "avg_speed_mph"]

peak_dur = peak_summary.loc["Peak", "avg_duration_min"]
offpeak_dur = peak_summary.loc["Off-Peak", "avg_duration_min"]

print("-" * 60)
print("OPERATIONAL BEHAVIORAL ANSWERS")
print("-" * 60)
print(f"• Total Peak Trips: {int(peak_trips):,} ({peak_trips_pct:.2f}% of overall fleet volume)")
print(f"• Total Off-Peak Trips: {int(offpeak_trips):,}")
print(f"• Speed Reduction during Peak: {offpeak_speed - peak_speed:.2f} mph slower ({peak_speed:.2f} mph vs {offpeak_speed:.2f} mph Off-Peak)")
print(f"• Duration Variance: Peak trips take average {peak_dur:.2f} min vs {offpeak_dur:.2f} min Off-Peak\n")
```

```python
import numpy as np
import pandas as pd

# Standardize payment_type labels if needed
if "payment_type_label" not in df_cleaned.columns:
    payment_map = {
        1: "Credit Card",
        2: "Cash",
        3: "No Charge",
        4: "Dispute",
        5: "Unknown",
    }
    df_cleaned["payment_type_label"] = df_cleaned["payment_type"].map(
        payment_map
    )

print("=" * 60)
print("11. TIP ANALYSIS - CORE METRICS & QUESTIONS")
print("=" * 60)

# 1. Average Tip (Overall & Credit Card Only)
avg_tip_overall = df_cleaned["tip_amount"].mean()
avg_tip_rate_overall = df_cleaned["tip_rate"].mean()

cc_df = df_cleaned[df_cleaned["payment_type_label"] == "Credit Card"]
avg_tip_cc = cc_df["tip_amount"].mean()
avg_tip_rate_cc = cc_df["tip_rate"].mean()

print(
    f"• Average Tip Amount (Overall Dataset): ${avg_tip_overall:.2f} (Tip Rate: {avg_tip_rate_overall:.2f}%)"
)
print(
    f"• Average Tip Amount (Credit Card Only): ${avg_tip_cc:.2f} (Tip Rate: {avg_tip_rate_cc:.2f}%)\n"
)

# 2. Highest Tip
max_tip = df_cleaned["tip_amount"].max()
max_tip_row = df_cleaned.loc[df_cleaned["tip_amount"].idxmax()]

print(f"• Highest Tip Amount Recorded: ${max_tip:.2f}")
print(
    f"  - Associated Fare: ${max_tip_row['fare_amount']:.2f} | Trip Distance: {max_tip_row['trip_distance']} miles | Payment: {max_tip_row['payment_type_label']}\n"
)

# 3. Credit Card vs Cash Tip Behavior
payment_tip_summary = df_cleaned.groupby("payment_type_label").agg(
    total_trips=("tip_amount", "count"),
    avg_tip_amount=("tip_amount", "mean"),
    avg_tip_rate=("tip_rate", "mean"),
    zero_tip_pct=("tip_amount", lambda x: (x == 0).mean() * 100),
)
print("• Credit Card vs. Cash Tipping Behavior:")
print(payment_tip_summary.round(2))
print()

# 4. Peak vs Off-Peak Tip Behavior
peak_tip_summary = df_cleaned.groupby("peak_period").agg(
    avg_tip_amount=("tip_amount", "mean"),
    avg_tip_rate=("tip_rate", "mean"),
    median_tip_rate=("tip_rate", "median"),
)
print("• Peak vs Off-Peak Tipping Behavior:")
print(peak_tip_summary.round(2))
print()

# 5. Long-Distance Trips Tip Rate
dist_tip_summary = df_cleaned.groupby("distance_category").agg(
    avg_tip_amount=("tip_amount", "mean"),
    avg_tip_rate=("tip_rate", "mean"),
    avg_fare=("fare_amount", "mean"),
)
print("• Tip Rates by Distance Category:")
print(dist_tip_summary.round(2))
print()

# 6. Passenger Count Tipping Behavior
pass_tip_summary = df_cleaned.groupby("passenger_group").agg(
    avg_tip_amount=("tip_amount", "mean"),
    avg_tip_rate=("tip_rate", "mean"),
    trip_count=("tip_amount", "count"),
)
print("• Tip Behavior by Passenger Group:")
print(pass_tip_summary.round(2))
print()

print("=" * 60)
print("CROSS-ANALYSIS TABLES (TIPPING BEHAVIOR)")
print("=" * 60)

# Cross 1: payment_type x tip_category
cross_pay_cat = pd.crosstab(
    df_cleaned["payment_type_label"],
    df_cleaned["tip_category"],
    normalize="index",
) * 100
print("\n1. payment_type × tip_category (% Share within Payment Type):")
print(cross_pay_cat.round(2))

# Cross 2: distance_category x avg tip_rate & tip_category distribution
cross_dist_rate = df_cleaned.groupby("distance_category")["tip_rate"].mean()
cross_dist_cat = pd.crosstab(
    df_cleaned["distance_category"],
    df_cleaned["tip_category"],
    normalize="index",
) * 100
print("\n2. distance_category × tip_category (% Distribution):")
print(cross_dist_cat.round(2))

# Cross 3: time_of_day x tip_rate
cross_time_rate = df_cleaned.groupby("time_of_day")["tip_rate"].agg(
    ["mean", "median"]
)
print("\n3. time_of_day × tip_rate:")
print(cross_time_rate.round(2))

# Cross 4: passenger_group x tip_rate
cross_pass_rate = df_cleaned.groupby("passenger_group")["tip_rate"].agg(
    ["mean", "median"]
)
print("\n4. passenger_group × tip_rate:")
print(cross_pass_rate.round(2))
```
